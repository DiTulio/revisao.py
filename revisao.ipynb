```python
#Exercício 1 - Crie uma função para calcular a soma, subtração, multiplicação e divisão para dois número. Retorne as 4 operações

def calcular(a, b):
    soma = a + b
    subtracao = a - b
    multiplicacao = a * b
    divisao = a / b if b != 0 else "Erro: divisão por zero"
    return soma, subtracao, multiplicacao, divisao

s, sub, mul, div = calcular(10, 2)
print(f"Soma: {s} | Subtração: {sub} | Multiplicação: {mul} | Divisão: {div}")
```

```python
#Exercício 2 - Crie uma função para transformar temperatura em fahrenheit para grau celsius

def fahrenheit_para_celsius():
    temp = float(input("Digite a temperatura em fahrenheit: "))
    celsius = (temp - 32) * 5 / 9
    print(f"{temp}°F = {celsius:.2f}°C")  # ✅ mostra os dois valores
    return celsius

resultado = fahrenheit_para_celsius()
```

```python
#Exercício 3 - Crie uma função a qual solicite o tipo de figura geométrica (círculo e retângulo). Após isso, crie um progresso programa para calcular a área da respectiva figura geométrica

import math

def calcular_area():
    figura = input("Digite o tipo de figura (circulo/retangulo): ").strip().lower()

    if figura == "circulo":
        raio = float(input("Digite o raio: "))
        area = math.pi * raio ** 2
        print(f"Área do círculo: {area:.2f}")

    elif figura == "retangulo":
        largura = float(input("Digite a largura: "))
        altura = float(input("Digite a altura: "))
        area = largura * altura
        print(f"Área do retângulo: {area:.2f}")

    else:
        print("Figura inválida! Escolha 'circulo' ou 'retangulo'.")

calcular_area()
```

```python
#Exercício 4 - Crie um função para calcular a média de 4 notaas de um aluno utilizando try/except. Ao final retorne a média deste aluno.

def calcular_media():
    try:
        nota1 = float(input("Digite a nota 1: "))
        nota2 = float(input("Digite a nota 2: "))
        nota3 = float(input("Digite a nota 3: "))
        nota4 = float(input("Digite a nota 4: "))

        media = (nota1 + nota2 + nota3 + nota4) / 4
        return media

    except ValueError:
        print("Erro: Digite apenas números para as notas.")
        return None

media = calcular_media()
if media is not None:
    print(f"Média do aluno: {media:.2f}")
```

```python
#Exercício 5 - Crie uma função para solicitar a palvra stop ou continue, caso o usuário digite stop, print ("Você deve parar."), caso digite continue, print("Você pode continuar."). Se o usuário não digitar nenhuma das duas palavras, a pergunta deve continuar de forma infinita, utilize estrutura while.

def stop_ou_continue():
    while True:
        resposta = input('Digite "stop" ou "continue": ').strip().lower()

        if resposta == "stop":
            print("Você deve parar.")
            break
        elif resposta == "continue":
            print("Você pode continuar.")
            break
        else:
            print("Palavra inválida! Tente novamente.\n")

stop_ou_continue()
```
